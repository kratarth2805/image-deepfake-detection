```python
def show_fft_sample(img_path):
    img = Image.open(img_path).convert("L")
    arr = np.array(img, dtype=np.float32)
    f = np.fft.fft2(arr)
    fshift = np.fft.fftshift(f)
    mag = 20 * np.log(np.abs(fshift) + 1e-8)
    
    plt.figure(figsize=(6, 3))
    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Spatial Image")
    plt.axis("off")
    
    plt.subplot(1, 2, 2)
    plt.imshow(mag, cmap="magma")
    plt.title("2D FFT Spectrum")
    plt.axis("off")
    plt.show()

# Run once dataset/processed contains images:
# show_fft_sample("dataset/processed/train/REAL/0000.jpg")
```
